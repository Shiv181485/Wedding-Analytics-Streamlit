```python
df.columns = (
    df.columns
      .str.strip()
      .str.replace(" ","_",regex=False)
      .str.replace("-","_",regex=False)
      .str.replace("/","_",regex=False)
      .str.replace("(","",regex=False)
      .str.replace(")","",regex=False)
)

print(df.columns.tolist())
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate Rows :",duplicate_rows)

df = df.drop_duplicates()

print("After Removing :",df.shape)
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numeric Columns")

print(numeric_cols)

print()

print("Categorical Columns")

print(categorical_cols)
```

```python
for col in df.columns:

    try:

        df[col]=pd.to_numeric(df[col])

    except:

        pass
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

categorical_cols = df.select_dtypes(exclude=np.number).columns

for col in numeric_cols:

    df[col].fillna(df[col].median(),inplace=True)

for col in categorical_cols:

    df[col].fillna(df[col].mode()[0],inplace=True)

print(df.isnull().sum())
```

```python
for col in categorical_cols:

    df[col]=df[col].astype(str).str.strip()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:

    Q1=df[col].quantile(.25)

    Q3=df[col].quantile(.75)

    IQR=Q3-Q1

    lower=Q1-1.5*IQR

    upper=Q3+1.5*IQR

    df=df[(df[col]>=lower)&(df[col]<=upper)]

print(df.shape)
```

```python
encoder=LabelEncoder()

for col in categorical_cols:

    try:

        df[col]=encoder.fit_transform(df[col])

    except:

        pass
```

```python
import pandas as pd
import numpy as np

# Make a copy of the original dataframe
ml_df = df.copy()

# Convert categorical columns into numbers
for col in ml_df.columns:
    if ml_df[col].dtype == "object":
        ml_df[col] = pd.factorize(ml_df[col])[0]

# Remove missing values
ml_df = ml_df.dropna()

print("Shape:", ml_df.shape)
print(ml_df.head())
```

```python
X = ml_df.drop("Cost_of_Type", axis=1)
y = ml_df["Cost_of_Type"]

print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,
    test_size=0.20,
    random_state=42

)

print(X_train.shape)

print(X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(

    n_estimators=300,

    random_state=42

)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

mae = mean_absolute_error(y_test, rf_pred)
rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
r2 = r2_score(y_test, rf_pred)

print("MAE :", round(mae,2))
print("RMSE :", round(rmse,2))
print("R² :", round(r2,3))
```

```python
import xgboost as xgb

xgb_model = xgb.XGBRegressor(

    n_estimators=500,

    learning_rate=0.05,

    max_depth=6,

    random_state=42

)

xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)
```

```python
rf_r2 = r2_score(y_test, rf_pred)

xgb_r2 = r2_score(y_test, xgb_pred)

comparison = pd.DataFrame({

"Model":[

"Random Forest",

"XGBoost"

],

"R2 Score":[

rf_r2,

xgb_r2

]

})

comparison
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# ================= KPIs ==================
total_weddings = len(df)
total_cost = df["Cost_of_Type"].sum()
avg_cost = df["Cost_of_Type"].mean()
max_cost = df["Cost_of_Type"].max()

fig = make_subplots(
    rows=1,
    cols=4,
    specs=[[{"type":"indicator"},
            {"type":"indicator"},
            {"type":"indicator"},
            {"type":"indicator"}]],
    horizontal_spacing=0.08
)

# ================= KPI 1 =================
fig.add_trace(
    go.Indicator(
        mode="number+delta",
        value=total_weddings,
        number={
            "font":{"size":52,"color":"#00F5FF"},
            "suffix":" 👰"
        },
        delta={
            "reference":total_weddings*0.90,
            "relative":True,
            "increasing":{"color":"#00FFB3"}
        },
        title={
            "text":"<b>Total Weddings</b>",
            "font":{"size":22,"color":"white"}
        },
        domain={'x':[0,0.23],'y':[0,1]}
    ),
    1,1
)

# ================= KPI 2 =================
fig.add_trace(
    go.Indicator(
        mode="number+gauge",
        value=total_cost,
        number={
            "prefix":"₹",
            "font":{"size":46,"color":"gold"}
        },
        title={
            "text":"<b>Total Budget</b>",
            "font":{"size":22,"color":"white"}
        },
        gauge={
            "shape":"bullet",
            "axis":{"visible":False},
            "bar":{"color":"gold"},
            "bgcolor":"#1f2937"
        }
    ),
    1,2
)

# ================= KPI 3 =================
fig.add_trace(
    go.Indicator(
        mode="number+delta",
        value=avg_cost,
        number={
            "prefix":"₹",
            "font":{"size":46,"color":"#39FF14"}
        },
        delta={
            "reference":avg_cost*0.85,
            "relative":True,
            "increasing":{"color":"#39FF14"}
        },
        title={
            "text":"<b>Average Cost</b>",
            "font":{"size":22,"color":"white"}
        }
    ),
    1,3
)

# ================= KPI 4 =================
fig.add_trace(
    go.Indicator(
        mode="number+gauge",
        value=max_cost,
        number={
            "prefix":"₹",
            "font":{"size":46,"color":"#FF4D6D"}
        },
        title={
            "text":"<b>Highest Wedding Cost</b>",
            "font":{"size":22,"color":"white"}
        },
        gauge={
            "shape":"bullet",
            "axis":{"visible":False},
            "bar":{"color":"#FF4D6D"},
            "bgcolor":"#1f2937"
        }
    ),
    1,4
)

# ================= Layout =================
fig.update_layout(

    template="plotly_dark",

    paper_bgcolor="#050816",
    plot_bgcolor="#050816",

    height=360,
    width=1650,

    title={
        "text":"<b>💎 EXECUTIVE WEDDING ANALYTICS DASHBOARD</b>",
        "x":0.5,
        "font":{
            "size":30,
            "color":"white",
            "family":"Arial Black"
        }
    },

    margin=dict(l=20,r=20,t=80,b=20),

    font=dict(
        color="white",
        family="Segoe UI"
    )
)

fig.show()
```
